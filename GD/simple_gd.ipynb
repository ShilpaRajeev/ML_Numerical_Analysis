```python
# simple vanilla gradient descent
import jax.numpy as jnp

def gradient_descent(grad_func, x0, lr=0.1, tol=1e-6, max_iter=1000):
    x = x0.copy()

    path = [x]
    for _ in range(max_iter): 
        grad_val = grad_func(x) 
        x = x - lr * grad_val
        path.append(x.copy()) 
        if jnp.linalg.norm(grad_val) < tol:
            break
    return x, path
# gradient function
def grad_func(x):
    return 2 * x

# initial point
x0 = jnp.array(5.0)

x_min, path = gradient_descent(grad_func, x0, lr=0.1)

print(x_min)
print(path[:5])   # first 5 iterations
```
