```python
import jax
# jax to compute gradient
import jax.numpy as jnp
import matplotlib.pyplot as plt

set
key = jax.random.PRNGKey(0)
```

```python
xx = jnp.linspace(-5, 5, 200)
yy = jnp.linspace(-5,5, 200)
X, Y = jnp.meshgrid(xx, yy)
Z = ((X + 2*Y - 7)**2 + (2*X +Y - 5)**2) #defining the function as Z

cs = plt.contourf(X,Y,Z) # takes the matrices and function
plt.colorbar(cs)
plt.plot([0,3,5],[1,2,2], "ro-")
```

```python
# Define grid
xx = np.linspace(-5, 5, 200)
yy = np.linspace(-5, 5, 200)
X, Y = np.meshgrid(xx, yy)

# Define the function
Z = (X + 2*Y - 7)**2 + (2*X + Y - 5)**2

# Contour plot
plt.figure(figsize=(8, 6))
cs = plt.contour(X, Y, Z, levels=30)
plt.clabel(cs, inline=True, fontsize=8)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Contour plot of f(x, y)")
plt.show()
```
